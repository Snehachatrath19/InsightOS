```python
import pandas as pd
```

```python
orders = pd.read_csv("../data/olist_orders_dataset.csv")
```

```python
order_items = pd.read_csv("../data/olist_order_items_dataset.csv")
```

```python
sales = pd.merge(
    orders,
    order_items,
    on="order_id"
)

sales.head()
```

```python
sales["order_purchase_timestamp"] = pd.to_datetime(
    sales["order_purchase_timestamp"]
)
```

```python
sales["Month"] = sales["order_purchase_timestamp"].dt.to_period("M")
```

```python
sales.head()
```

```python
monthly_revenue = (
    sales.groupby("Month")["price"]
    .sum()
    .reset_index()
)

monthly_revenue.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(monthly_revenue["Month"].astype(str),
         monthly_revenue["price"],
         marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

## Business Insight

### Observation
Revenue increased steadily during the observed period, indicating business growth.

### Possible Reasons
- Increased customer acquisition
- Better marketing campaigns
- More products sold

### Recommendation
Investigate the highest-performing months and identify the strategies that contributed to their success. Apply those strategies during lower-performing months.

```python
products = pd.read_csv("../data/olist_products_dataset.csv")

products.head()
```

```python
sales_products = pd.merge(
    order_items,
    products,
    on="product_id"
)

sales_products.head()
```

```python
top_products = (
    sales_products
    .groupby("product_id")["price"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
top_quantity = (
    sales_products
    .groupby("product_id")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

top_quantity
```

```python
import matplotlib.pyplot as plt

top_products.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Top 10 Revenue Generating Products")
plt.xlabel("Product ID")
plt.ylabel("Revenue ($)")
plt.xticks(rotation=90)

plt.show()
```

## Business Insight

The top 10 products contribute a significant portion of total revenue.

Recommendation:
- Maintain inventory for these products.
- Promote them during seasonal campaigns.
- Analyze why lower-performing products are underperforming.
