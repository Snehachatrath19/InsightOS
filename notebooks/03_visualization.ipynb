# 📊 Module 3 - Data Visualization

In this notebook we create interactive business dashboards using Plotly.

```python
import pandas as pd
import plotly.express as px
```

```python
orders = pd.read_csv("../data/olist_orders_dataset.csv")
orders.head()
```

```python
order_items = pd.read_csv("../data/olist_order_items_dataset.csv")
order_items.head()
```

```python
sales = pd.merge(
    orders,
    order_items,
    on="order_id"
)

sales.head()
```

```python
status = sales["order_status"].value_counts().reset_index()
status.columns = ["Status", "Orders"]

fig = px.bar(
    status,
    x="Status",
    y="Orders",
    title="Orders by Status"
)

fig.show()
```

```python
status = sales["order_status"].value_counts().reset_index()

status.columns = ["Status", "Orders"]

fig = px.bar(
    status,
    x="Status",
    y="Orders",
    title="Orders by Status",
    color="Status"
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px
```

```python
status = sales["order_status"].value_counts().reset_index()

status.columns = ["Status", "Orders"]

fig = px.bar(
    status,
    x="Status",
    y="Orders",
    title="Orders by Status",
    color="Status"
)

fig.show()
```
